### Arithmetic Operators in Python

```python
print(10 + 3)   # Addition → 13
print(10 - 3)   # Subtraction → 7
print(10 * 3)   # Multiplication → 30
print(10 / 3)   # Division → 3.333...
print(10 // 3)  # Floor division → 3
print(10 % 3)   # Remainder (modulus) → 1
print(10 ** 3)  # Exponentiation → 1000
```

### Comparison Operators

```python
print(10 == 10)  # Equal to → True
print(10 != 3)   # Not equal to → True
print(10 > 3)    # Greater than → True
print(10 < 3)    # Less than → False
print(10 >= 10)  # Greater than or equal to → True
print(10 <= 3)   # Less than or equal to → False
```

### Logical Operators

```python
print(10 > 5 and 10 < 20)  # Both conditions are True → True
print(10 < 5 or 10 < 20)   # At least one condition is True → True
print(not 10 > 5)           # Reverses True to False → False
```

### Assignment Operators

```python
number = 10      # Assign 10 to number
number += 5      # Same as number = number + 5 → 15
number -= 3      # Same as number = number - 3 → 12
number *= 2      # Same as number = number * 2 → 24
number /= 4      # Same as number = number / 4 → 6.0
number //= 3      # Same as number = number // 3 → 2.0
print(number)
```

### Membership Operator

```python
print("AI" in "Programming for AI")         # Found → True
print("Java" not in "Programming for AI")   # Not found → True
```

```python
numbers = [10, 20, 30, 40]

print(20 in numbers)      # 20 exists in the list → True
print(50 not in numbers)  # 50 does not exist in the list → True
```

## Identity Operator

```python
a = [10, 20]
b = [10, 20]

print(a == b)  # Same values → True
print(a is b)  # Different objects → False
```

### Student Marks Calculator

```python
quiz = float(input("Quiz: "))
assignment = float(input("Assignment: "))
mid = float(input("Mid: "))
final = float(input("Final: "))
 
maximum_marks = 100
total = quiz + assignment + mid + final
percentage = (total / maximum_marks) * 100
print("Total:", total, "/", maximum_marks)
print("Percentage:", percentage)
```
