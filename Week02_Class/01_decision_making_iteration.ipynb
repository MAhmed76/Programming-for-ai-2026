#### Admission Eligibility System

```python
matric = float(input("Matric %: "))
intermediate = float(input("FSC %: "))
test = float(input("Entry Test %: "))
 
score = (matric + intermediate + test) / 3
 
if score >= 80:
    status = "Eligible"
elif score >= 60:
    status = "Waiting list"
else:
    status = "Rejected"
 
print(status)
```

```python
# Practice: Transform and Filter Model Scores

scores = [0.72, 0.91, 0.83, 0.65, 0.95]
high_scores = []
for score in scores:
    if score >= 0.80:
        high_scores.append(score * 100)

print(high_scores)
```

```python
high_scores = [score * 100 for score in scores if score >= 0.80]
print(high_scores)
```

#### Program Flow

```python
# Password validator
password = input("Enter a password: ")

if len(password) >= 8:
    print("Valid password")
else:
    print("Password is too short")
```

```python
# Guess the Number Game
secret_number = 7
guess = 0

while guess != secret_number:
    guess = int(input("Guess the number: "))

    if guess < secret_number:
        print("Too low")
    elif guess > secret_number:
        print("Too high")

print("Correct!")
```

```python
# Temperature Monitoring System
for i in range(3):
    temperature = float(input("Enter temperature: "))

    if temperature > 30:
        print("High temperature")
    elif temperature < 15:
        print("Low temperature")
    else:
        print("Normal temperature")
```
